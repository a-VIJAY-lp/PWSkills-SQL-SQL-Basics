```python
import pandas as pd

df = pd.read_csv('Retail_Dataset.csv')
```

```python
#Q2. Data Inspection
#Perform the following inspection tasks:
#1) Display the first 5 rows of the dataset.
#2) Display the last 5 rows.
#3) Show the structure of the dataset using info().
#4) Display the shape of the dataset
# First 5 rows
print(df.head())

# Last 5 rows
print(df.tail())

# Structure of the dataset
print(df.info())

# Shape of the dataset
print(df.shape)
```

```python
#Q3. Column Selection
#Select and display only the following columns:
#Product
#Region
#Sales
selected_cols = df[['Product', 'Region', 'Sales']]
print(selected_cols)
```

```python
#Q4. Indexing using .loc
#Using .loc, perform the following:
#Display the row with index 10.
#Display the Sales value of the row with index 20
# Row with index 10
row_10 = df.loc[10]
print('Row with index 10')
print(row_10)

# Sales value of row with index 20
sales_20 = df.loc[20, 'Sales']

print(f"'Sales value of row with index 20': {sales_20}")
```

```python
#Q5. Indexing using .iloc
#Using .iloc, display:
#1) The first 6 rows and first 4 columns.
#2) The Sales value from the 3rd row.
# First 6 rows and first 4 columns
first6_first4 = df.iloc[0:6, 0:4]
print(first6_first4)

# Sales value from the 3rd row
sales_3rd_row = df.iloc[2, df.columns.get_loc('Sales')]
print(f" Sales value from the 3rd row: {sales_3rd_row}")
```

```python
#Q6. Row Slicing
#Display: The first 10 rows of the dataset.Rows 15 to 25.
# First 10 rows
first_10 = df.iloc[0:10]
print(first_10)

# Rows 15 to 25 (15 included, 26 excluded)
rows_15_25 = df.iloc[15:26]
print(rows_15_25)
```

```python
#Q7. Filtering Data
#Filter the dataset to display:
#1) Rows where Region = "East".
#2) Rows where Sales > 1500.
# Region = "East"
east_rows = df[df['Region'] == 'East']
print(east_rows)

# Sales > 1500
high_sales = df[df['Sales'] > 1500]
print(high_sales)
```

```python
#Q8. Multiple Condition Filtering
#Display rows where:
#Region is West
#Sales is greater than 1200
west_high_sales = df[(df['Region'] == 'West') & (df['Sales'] > 1200)]
print(west_high_sales)
```

```python
#Q9. Descriptive Statistics
#Generate summary statistics using:
#describe()
#From the result identify:
#Mean sales value
#Maximum quantity sold
#Minimum sales value
stats = df.describe()
print(stats)

# Mean sales value
mean_sales = stats.loc['mean', 'Sales']
print("Mean Sales:", mean_sales)

# Maximum quantity sold
max_quantity = stats.loc['max', 'Quantity']
print("Maximum Quantity:", max_quantity)

# Minimum sales value
min_sales = stats.loc['min', 'Sales']
print("Minimum Sales:", min_sales)
```

```python
#Q10. Aggregation Analysis
#Using the Sales column, calculate:
#Total sales
#Average sales
#Maximum sales
#Minimum sales.
total_sales = df['Sales'].sum()
avg_sales = df['Sales'].mean()
max_sales = df['Sales'].max()
min_sales = df['Sales'].min()

print("Total Sales:", total_sales)
print("Average Sales:", avg_sales)
print("Maximum Sales:", max_sales)
print("Minimum Sales:", min_sales)
```
