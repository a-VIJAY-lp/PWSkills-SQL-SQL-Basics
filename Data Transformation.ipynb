```python
import pandas as pd

df = pd.read_csv("employee_productivity_dataset.csv")
df.head()
```

```python
#Q1. Handle Missing Values (Basic Cleaning)
#Fill missing values using the following:
    #Age → Median
    #Salary → Mean
    #Hours_Worked_Per_Week → Median
    #Performance_Score → Mean
#Display the dataset after handling missing values.
# 1. Check missing values
df.isna().sum()

# 2. Compute required statistics
age_median = df["Age"].median()
salary_mean = df["Salary"].mean()
hours_median = df["Hours_Worked_Per_Week"].median()
perf_mean = df["Performance_Score"].mean()

#alternative appraoch
#df = df.fillna({
#    "Age": age_median,
#    "Salary": salary_mean,
#    "Hours_Worked_Per_Week": hours_median,
#    "Performance_Score": perf_mean
#})

# 3. Fill missing values
df["Age"] = df["Age"].fillna(age_median)
df["Salary"] = df["Salary"].fillna(salary_mean)
df["Hours_Worked_Per_Week"] = df["Hours_Worked_Per_Week"].fillna(hours_median)
df["Performance_Score"] = df["Performance_Score"].fillna(perf_mean)

# 4. Display dataset after cleaning
df.head()
df.isna().sum()
```

```python
#Q2. Label Encoding
#Convert these columns into numeric values using Label Encoding:
#Gender
#Department
#Show the updated columns
from sklearn.preprocessing import LabelEncoder
print(df[["Gender", "Department"]].head())
le_gender = LabelEncoder()
le_dept = LabelEncoder()

df["Gender"] = le_gender.fit_transform(df["Gender"])
df["Department"] = le_dept.fit_transform(df["Department"])

# Show updated columns
df[["Gender", "Department"]].head()
```

```python
#Q3. One-Hot Encoding
#Apply One-Hot Encoding on:
#Work_Mode
#Location
#Display the dataset and check how many new columns are created.
print(df.head())
print("-"*100)
# 1. Apply one‑hot encoding
df_encoded = pd.get_dummies(
    df,
    columns=["Work_Mode", "Location"],
    drop_first=False  # keep all categories
)

# 2. Display dataset
print(df_encoded.head())

# 3. Check how many new columns were created
new_cols = [c for c in df_encoded.columns if ("Work_Mode_" in c) or ("Location_" in c)]
len_new_cols = len(new_cols)
print("\n",new_cols, len_new_cols)
```

```python
#Q4. Normalization (Min-Max Scaling)
#Normalize the following columns:
#Salary
#Hours_Worked_Per_Week
#Use MinMaxScaler and display results
from sklearn.preprocessing import MinMaxScaler

scaler_minmax = MinMaxScaler()

cols_norm = ["Salary", "Hours_Worked_Per_Week"]

df_minmax = df.copy()
df_minmax[cols_norm] = scaler_minmax.fit_transform(df_minmax[cols_norm])

df_minmax[cols_norm].head()
```

```python
#Q5. Standardization (Scaling)
#Apply StandardScaler on:
#Age
#Projects_Completed
#Display the transformed values.
from sklearn.preprocessing import StandardScaler

scaler_std = StandardScaler()

cols_std = ["Age", "Projects_Completed"]

df_std = df.copy()
df_std[cols_std] = scaler_std.fit_transform(df_std[cols_std])

df_std[cols_std].head()
```

```python
#Q6. Compare Scaling Methods
#Apply both:
#MinMaxScaler
#StandardScaler
#On the Salary column.
#Show both results side by side

from sklearn.preprocessing import MinMaxScaler, StandardScaler

salary_values = df[["Salary"]]  # as 2D array

mm = MinMaxScaler()
ss = StandardScaler()

salary_minmax = mm.fit_transform(salary_values)
salary_std = ss.fit_transform(salary_values)

salary_compare = pd.DataFrame({
    "Salary_Original": salary_values["Salary"].values,
    "Salary_MinMax": salary_minmax.flatten(),
    "Salary_Standard": salary_std.flatten()
})

salary_compare.head()
```

```python
#Q7. Build Preprocessing Pipeline
#Create a pipeline that:
#Applies encoding to categorical columns
#Applies scaling to numerical columns
#Use:
#ColumnTransformer
#Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

# 1. Define column groups
cat_cols = ["Gender", "Department", "Work_Mode", "Location"]
num_cols = ["Age", "Salary", "Hours_Worked_Per_Week",
            "Performance_Score",  "Projects_Completed"]

# 2. Define transformers
cat_transformer = OneHotEncoder(handle_unknown="ignore")
num_transformer = StandardScaler()

# 3. ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", cat_transformer, cat_cols),
        ("num", num_transformer, num_cols)
    ]
)

# 4. Build pipeline (only preprocessing, no model)
preprocess_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor)
])
```

```python
#Q8. Apply Pipeline
#Apply the pipeline on the dataset.
#Display:
#Transformed dataset
#Shape of final dataset
# 1. Fit and transform the dataset
X_transformed = preprocess_pipeline.fit_transform(df)

# 2. Convert to array or DataFrame
X_transformed.shape

# Optional: make DataFrame with generated column names
ohe = preprocess_pipeline.named_steps["preprocessor"].named_transformers_["cat"]
cat_feature_names = ohe.get_feature_names_out(cat_cols)

all_feature_names = list(cat_feature_names) + num_cols

import numpy as np
X_df = pd.DataFrame(X_transformed.toarray() if hasattr(X_transformed, "toarray") else X_transformed,
                    columns=all_feature_names)

X_df.head()
X_df.shape
```

#Q9. Why is scaling important in Python?
Scaling is important because:

Many machine learning algorithms (like KNN, SVM, logistic regression, neural networks) are sensitive to feature magnitude; unscaled features can dominate the learning process.

It helps gradients converge faster during optimization and avoids numerical problems when features have very different ranges.

Distance‑based methods compute distances in feature space; if you scale features, each feature contributes more fairly to the distance.

Q10. Why convert categorical data to numeric?
We convert categorical data into numerical form because:

Most machine learning algorithms work only with numerical inputs and cannot directly process text categories.

Encoding categories into numbers preserves useful information while allowing models to compute patterns, distances, probabilities, and decision boundaries.

Proper encodings (like one‑hot) avoid introducing fake order and make training more stable and interpretable
