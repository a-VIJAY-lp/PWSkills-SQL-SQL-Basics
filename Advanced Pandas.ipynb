```python
import pandas as pd
df = pd.read_csv("Advanced_pandas_sales_dataset.csv")   # change name/extension if needed
```

Q2 Sales Filtering Analysis
    Filter the dataset to display:
    Orders where Sales > 3500
    Orders where Region = "East" and Sales > 2500
    Explain what insights you observe.

```python
# Q2 - Sales Filtering Analysis

# 1) Orders where Sales > 3500
high_sales = df[df["Sales"] > 3500]
print("Orders where Sales > 3500:")
print(high_sales)
```

```python
# 2) Orders where Region = 'East' and Sales > 2500
east_high_sales = df[(df["Region"] == "East") & (df["Sales"] > 2500)]
print("\nOrders where Region = 'East' and Sales > 2500:")
print(east_high_sales)
```

```python
count_high_sales=high_sales.shape[0];
print("Number of orders with Sales > 3500:", count_high_sales)

print("\n" + "-"*100 + "\n")

count_east_high = east_high_sales.shape[0]
print("Number of East orders with Sales > 2500:", count_east_high)

print("\n" + "-"*100 + "\n")

# Top products contributing to higher revenue  (Sales > 3500)
top_products = (
    high_sales.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
print("Top products  (Sales > 3500, by total Sales):")
print(top_products)
print("\n" + "-"*100 + "\n")
# Top products contributing to higher revenue in East (Sales > 2500)
top_products_east = (
    east_high_sales.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
print("Top products in East (Sales > 2500, by total Sales):")
print(top_products_east)

print("\n" + "-"*100 + "\n")

# Top customers contributing to higher revenue (Sales > 3500)
top_customers = (
    high_sales.groupby("Customer")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
print("Top customers (Sales > 3500, by total Sales):")
print(top_customers)
print("\n" + "-"*100 + "\n")
# Top customers contributing to higher revenue in East (Sales > 2500)
top_customers_east = (
    east_high_sales.groupby("Customer")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
print("Top customers in East (Sales > 2500, by total Sales):")
print(top_customers_east)
```

```python
#Q3 Customer Purchase Analysis
#Identify the top 5 customers based on total sales.
#Hint: Use GroupBy with aggregation.

# Group by customer and calculate total sales
customer_sales = (
    df.groupby("Customer")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Total sales by customer (descending):")
print(customer_sales)

print("\nTop 5 customers based on total sales:")
top_5_customers = customer_sales.head(5)
print(top_5_customers)
```

```python
#Q4 Regional Sales Performance
#Calculate the total sales generated by each region.
#Then identify:
#Which region generated the highest revenue?
# Q4 - Regional Sales Performance

# 1) Total sales generated by each region
regional_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Total sales by region (descending):")
print(regional_sales)

print("\nRegion with the highest revenue:")
top_region = regional_sales.idxmax()
top_region_value = regional_sales.max()
print(f"{top_region} with total sales of {top_region_value}")
```

```python
#Q5 Product Performance Analysis
#Calculate the total quantity sold for each product.
#Which product appears to be the most popular?


# 1) Total quantity sold for each product
product_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("Total quantity sold for each product (descending):")
print(product_quantity)

print("\nMost popular product (by quantity sold):")
most_popular_product = product_quantity.idxmax()
most_popular_quantity = product_quantity.max()
print(f"{most_popular_product} with total quantity {most_popular_quantity}")
```

```python
#Q6 Multi-Level Grouping
#Group the dataset by: Region and Category
#Calculate the total sales for each combination.
#What insights can you derive from this analysis?


multi_group_sales = (
    df.groupby(["Region", "Category"])["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Total sales for each (Region, Category) combination:")
print(multi_group_sales)

multi_group_sales_df = multi_group_sales.reset_index()
print("\nAs a DataFrame:")
print(multi_group_sales_df)

print("\n ### 1. Best region–category combination \n - The **highest** sales are in the **East–Electronics** segment with 160,907, making it your strongest performing combination.\n - This suggests Electronics in the East should be a **priority** for marketing, inventory, and customer retention efforts.\n\n### 2. Category performance across regions\n - **Electronics** leads in every region compared to Accessories and Office, showing it is the overall strongest category.\n - Among Electronics, the region ranking is: East (160,907) > South (86,690) > West (84,128) > North (70,023). \n\n### 3. Accessories vs Office supplies \n - Accessories consistently outperform Office supplies in all regions (for example, South Accessories 80,914 vs South Office 15,504). \n - Office supplies are the **weakest** category overall, so they may need targeted promotions or could be a lower strategic focus. \n\n ### 4. Region-level hints \n - The **South** and **West** regions show strong performance in both Electronics and Accessories, indicating good diversified demand. \n - The **North** region has lower values overall, especially for Office (26,135), suggesting it’s a potential area for growth. ")
```

```python
#Q7 Pivot Table Creation
#Create a pivot table showing:
#Rows → Region
#Columns → Product
#Values → Total Sales
#Which region sells the most laptops?


pivot_q7 = pd.pivot_table(
    df,
    index="Region",        # rows
    columns="Product",     # columns
    values="Sales",        # values to aggregate
    aggfunc="sum",         # total sales
    fill_value=0           # replace NaN with 0
)

print("Pivot table: Total Sales by Region and Product")
print("-"*100)
print(pivot_q7)
print("-"*100)
print("Region with highest laptop sales")
print("\nLaptop column values by region: East 55,568; South 23,029; West 10,826; North 4,540.So, the East region has the highest laptop sales (55,568), making it the strongest market for laptops.")
print("\nEast also shows high sales in Phones (39,667) and Tablets (34,373), indicating a generally strong electronics demand there.")
print("\n\n \"According to the pivot table, the East region records the highest laptop sales (55,568), so it is the top-performing region for laptops.\"")
```

```python
#Q8 Profitability Analysis
#Calculate the average profit for each product category.
#Which category is most profitable?


# 1) Average profit for each product category
avg_profit_by_category = (
    df.groupby("Category")["Profit"]
    .mean()
    .sort_values(ascending=False)
)

print("Average profit for each category (descending):")
print(avg_profit_by_category)

# 2) Most profitable category
most_profitable_category = avg_profit_by_category.idxmax()
highest_avg_profit = avg_profit_by_category.max()

print(f"\nMost profitable category: {most_profitable_category} (avg profit: {highest_avg_profit})")
```

```python
#Q9 Discount Impact Study
#Analyze the relationship between discount and profit.
#Tasks:
#Calculate the average profit for each discount level
#Identify whether higher discounts lead to lower profit

# 1) Average profit for each discount level
avg_profit_by_discount = (
    df.groupby("Discount")["Profit"]
    .mean()
    .sort_index()
)

print("Average profit for each discount level:")
print(avg_profit_by_discount)
print("-"*100)
# Rough check: correlation between Discount and Profit
corr = df["Discount"].corr(df["Profit"])
print(f"\nCorrelation between Discount and Profit: {corr}")

print("Insight from average profit by discount")
print("-"*40)
print("\nAverage profit drops steadily as discount increases:0.00 → 751.21, 0.05 → 531.37, 0.10 → 346.46, 0.15 → 237.19, 0.20 → 90.03")

print("\nThis shows that higher discounts clearly reduce profit per order, and full-price (0% discount) sales are the most profitable.")

print("\n\nInsight from correlation")
print("-"*25)
print("\nThe correlation between Discount and Profit is -0.664, which is a strong negative relationship.")
print("\nThis means that as Discount goes up, Profit tends to go down, confirming that aggressive discounting harms profitability in this dataset.")
print("\n“The analysis shows a clear downward trend in average profit as discount levels increase, and the strong negative correlation (-0.66) between discount and profit confirms that higher discounts significantly reduce profitability.“")
```

```python
#Q10 End-to-End Data Wrangling Task
#Perform the following workflow:
#Filter the dataset where Sales > 3000
#Group data by Region and Product
#Calculate total sales
#Reshape the result into a pivot table
#Explain the business insights obtained from this analysis


# 1) Filter the dataset where Sales > 3000
filtered_df = df[df["Sales"] > 3000]
print("Filtered rows (Sales > 3000):")
print(filtered_df.head())
print("-"*100)
# 2) Group by Region and Product, calculate total sales
grouped_q10 = (
    filtered_df.groupby(["Region", "Product"])["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTotal sales by (Region, Product) after filtering:")
print(grouped_q10)

print("-"*100)
# 3) Reshape the result into a pivot table
pivot_q10 = grouped_q10.reset_index().pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)

print("\nPivot table: Total Sales (Sales > 3000) by Region and Product")
print(pivot_q10)
print("-"*100)
# 4)Explain the business insights obtained from this analysis
print("East region generates the highest premium laptop sales (44,096), while West dominates high-value keyboard sales (33,380). South shows strong high-value demand for mice (26,928), whereas North has more balanced high-value sales across multiple products. This suggests that product strategy can be region-specific: laptops and tablets in East, keyboards in West, and accessories in South and North.")
```
