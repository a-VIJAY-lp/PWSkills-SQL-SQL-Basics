```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('House Price India.csv')
df.head()
df.info()
df.describe()
```

```python
#1) Univariate analysis (distributions, central tendency, spread)
num_cols = ['number of bedrooms','number of bathrooms','living area','lot area',
            'number of floors','number of views','condition of the house',
            'grade of the house','Area of the house(excluding basement)',
            'Area of the basement','Built Year','Renovation Year',
            'Number of schools nearby','Distance from the airport','Price']

for col in num_cols:
    print('\n\nColumn:', col)
    print(df[col].describe())   # count, mean, std, min, quartiles, max
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
#For categorical/binary:
cat_cols = ['waterfront present','Postal Code']

for col in cat_cols:
    print('\n\nColumn:', col)
    print(df[col].value_counts())
    plt.figure(figsize=(6,4))
    sns.countplot(x=df[col])
    plt.title(f'Countplot of {col}')
    plt.xticks(rotation=45)
    plt.show()
```

```python
#Bivariate analysis (relationships with Price)
#Numeric vs Price (scatter + correlation):
for col in num_cols:
    if col == 'Price':
        continue
    plt.figure(figsize=(6,4))
    sns.scatterplot(x=df[col], y=df['Price'])
    plt.title(f'{col} vs Price')
    plt.show()

corr = df[num_cols].corr()
corr['Price'].sort_values(ascending=False)
```

```python
#Categorical vs Price (boxplots):
plt.figure(figsize=(6,4))
sns.boxplot(x='number of bedrooms', y='Price', data=df)
plt.title('Price by Number of Bedrooms')
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x='waterfront present', y='Price', data=df)
plt.title('Price by Waterfront Presence')
plt.show()

plt.figure(figsize=(8,4))
sns.boxplot(x='grade of the house', y='Price', data=df)
plt.title('Price by House Grade')
plt.xticks(rotation=45)
plt.show()
```

```python
#3) Missing values & data quality
df.isnull().sum()
print(df.isnull().sum())
# Example: drop rows with missing Price, impute others
df = df.dropna(subset=['Price'])
df['number of bathrooms'] = df['number of bathrooms'].fillna(df['number of bathrooms'].median())
print(df)
```

```python
#Q1. What is the average, median, and standard deviation of house prices?
mean_price = df['Price'].mean()
median_price = df['Price'].median()
std_price = df['Price'].std()

print(f"Mean Price : {mean_price}, Median Price : {median_price}, Standard Deviation : {std_price}")

print("\nThe mean price shows the typical market value considering all houses.")

print("\nThe median is less affected by extreme high‑priced properties, so it represents the central tendency more robustly.")

print("\nThe standard deviation indicates how widely prices vary around the average: a higher value means large price dispersion across houses.")
```

```python
#Q2. Which number of bedrooms is most common?
# Frequency of each bedroom count
df['number of bedrooms'].value_counts()

# Most common (mode)
most_common_bedrooms = df['number of bedrooms'].mode()[0]
print(most_common_bedrooms)
print (f"This suggests that {most_common_bedrooms}‑bedroom houses are the most typical/standard configuration in this market.")
```

```python
#Q3. Check if the price data is skewed.
import seaborn as sns
import matplotlib.pyplot as plt

# 1) Visual check with histogram
plt.figure(figsize=(6,4))
sns.histplot(df['Price'], kde=True)
plt.title('Distribution of House Prices')
plt.show()

# 2) Numerical skewness
price_skew = df['Price'].skew()
print(f"price skew : {price_skew}")
```

```python
#Q4. What is the average price for each number of bedrooms?
# Average price grouped by number of bedrooms
avg_price_by_bed = df.groupby('number of bedrooms')['Price'].mean().sort_index()
print(f"Average price per bedroom : {avg_price_by_bed}")

avg_price_by_bed_df = avg_price_by_bed.reset_index()
avg_price_by_bed_df.columns = ['number of bedrooms', 'average price']
print(avg_price_by_bed_df)
```

```python
#Q5.  What is the relationship between living area and price? (Use : Area without Basement).
import seaborn as sns
import matplotlib.pyplot as plt

# Scatter plot: living area (without basement) vs price
plt.figure(figsize=(6,4))
sns.scatterplot(
    x=df['Area of the house(excluding basement)'],
    y=df['Price']
)
plt.title('Area (excluding basement) vs Price')
plt.xlabel('Area of the house (excluding basement)')
plt.ylabel('Price')
plt.show()

# Correlation between area (excluding basement) and price
corr_area_price = df['Area of the house(excluding basement)'].corr(df['Price'])
print(corr_area_price)
print("The scatter plot shows a positive relationship between living area (excluding basement) and price: as the living area increases, house prices tend to increase.")
print("The Pearson correlation between living area (excluding basement) and price is about 0.62, indicating a moderately strong positive relationship: as the usable living area increases, house prices generally increase.")
```

```python
#Q6. Identify any anomalies where houses have high prices but low areas
area_col = 'Area of the house(excluding basement)'
price_col = 'Price'

# Check basic stats
print(df[[area_col, price_col]].describe())

q1_area = df[area_col].quantile(0.25)   # low area threshold
q3_price = df[price_col].quantile(0.75) # high price threshold

anomalies = df[(df[area_col] < q1_area) & (df[price_col] > q3_price)]
print(anomalies[[area_col, price_col]].head())
print(len(anomalies))
anomalies.head(10)
print("\n\nLow area was defined as below the 25th percentile of area (X sq.ft), and high price as above the 75th percentile of price (Y units).")

print(f"\n\nUsing this rule, we identified {len(anomalies)} houses that have unusually high prices despite low living area. These are potential anomalies and may be due to factors such as premium locations, special features (e.g., waterfront), or data entry issues.")
```

```python
#Q7. Compare average price based on number of floors and houses with or without waterfront.
avg_price_by_floors = df.groupby('number of floors')['Price'].mean().sort_index()
print(avg_price_by_floors)
avg_price_by_waterfront = df.groupby('waterfront present')['Price'].mean()
avg_price_by_waterfront
print("\n\n\n“Within each floor category, properties with waterfront tend to be more expensive than those without, showing a consistent waterfront premium across building heights.")
```

```python
#Q8. Identify the minimum and maximum house price. What does this indicate?
min_price = df['Price'].min()
max_price = df['Price'].max()

min_price, max_price
print(f"\nThe minimum house price in the dataset is {min_price}.")

print(f"\nThe maximum house price in the dataset is {max_price}.")
```

```python
#Q9. Which location (zipcode/area) has the highest average price?
# Average price by Postal Code
avg_price_by_zip = df.groupby('Postal Code')['Price'].mean().sort_values(ascending=False)
avg_price_by_zip.head()         # top few

top_zip = avg_price_by_zip.index[0]
top_zip_avg_price = avg_price_by_zip.iloc[0]
print(f" {top_zip} zip code is top average price and is  {top_zip_avg_price}")
```

```python
#Q10. Write at least 5 insights from your analysis.
print("\n\n1.Price distribution & spread")
min_price = df['Price'].min()
max_price = df['Price'].max()
mean_price = df['Price'].mean()
median_price = df['Price'].median()
std_price = df['Price'].std()

print(min_price, max_price, mean_price, median_price, std_price)
print("The minimum house price in the dataset is about ₹78,000, while the maximum is about ₹77,00,000, indicating a very wide price range in the market.")

print("\n\nThe average (mean) price is approximately ₹5,38,932, whereas the median price is about ₹4,50,000; since the mean is higher than the median, the distribution of prices is right‑skewed, driven by a few very expensive properties.")

print("\n\nThe standard deviation of prices is roughly ₹3,67,532, showing substantial variation around the average, which confirms a highly dispersed price distribution across houses.")
print("\nHouse prices are right‑skewed, with most homes in the low–mid range and a few very expensive properties, indicating a market with some luxury outliers.")

print("\n\n2.Bedrooms & price")
print(df['number of bedrooms'].mode()[0])
avg_price_by_bed = df.groupby('number of bedrooms')['Price'].mean().sort_index()
print(avg_price_by_bed)
print("\n The most common configuration is 3‑bedroom houses, indicating that 3BHK is the standard offering in this market.")
print("\n\n Average prices rise as bedrooms increase: about ₹3.09 lakh for 1 bedroom, ₹3.99 lakh for 2, ₹4.63 lakh for 3, ₹6.36 lakh for 4, and up to roughly ₹12.08 lakh for 8 bedrooms.")
print("\n\n This pattern shows a clear positive relationship between number of bedrooms and price: buyers pay a progressively higher premium for additional bedrooms, reflecting demand for larger family‑sized houses.")

print("\n\n3. Living area (excluding basement) and price")
print("\n\n correlation: 0.6152204198678111.")
print("\n\n There is a moderately strong positive correlation of about 0.62 between living area (excluding basement) and house price, meaning that larger usable floor area is generally associated with higher prices.")

print("\n\n The scatter plot of area (excluding basement) versus price shows an upward trend, with higher‑area houses clustering at higher price levels, confirming that floor area is a key driver of value in this market.")
print("There is a moderately strong positive correlation (~0.62) between living area (excluding basement) and price, meaning larger usable area usually leads to higher prices.")

print("4.Floors & waterfront effect")
df.groupby('number of floors')['Price'].mean().sort_index()
df.groupby('waterfront present')['Price'].mean()
df.groupby('number of floors')['Price'].mean().sort_index()
print("\n\nAverage price increases with the number of floors: about ₹4.37 lakh for 1 floor, ₹5.55 lakh for 1.5 floors, ₹6.49 lakh for 2 floors, and over ₹11.06 lakh for 2.5 floors, with 3.5‑floor houses reaching around ₹12.06 lakh. This shows that taller houses with more usable vertical space tend to be valued higher.")
print("\n\n Waterfront has a very strong price effect: non‑waterfront houses (0) average about ₹5.30 lakh, whereas waterfront houses (1) average roughly ₹16.42 lakh, i.e., around three times higher. This indicates a clear waterfront premium, meaning proximity to water is a major driver of house prices in this dataset.")



#Location impact (Postal Code)
avg_price_by_zip = df.groupby('Postal Code')['Price'].mean().sort_values(ascending=False)
avg_price_by_zip.head()

print("\n\nThe postal code 122071 has the highest average house price at about ₹23.48 lakh, followed by 122048 (~₹12.96 lakh) and 122057 (~₹11.89 lakh), which are all much higher than most other areas.")
print("\n\nThis pattern shows that location strongly influences house prices: certain postal codes such as 122071 form clear premium localities, where buyers pay substantially more compared to other regions in the dataset.")

```
