```python
import numpy as np
import pandas as pd

# Q1: Load the dataset
df = pd.read_csv("retail_dataset .csv")   

# Show first 5 rows
display(df.head())

# Show dataset information
df.info()
```

```python
# Q2: Missing values per column
missing_per_column = df.isnull().sum()
print("Missing values per column:")
print(missing_per_column)

# Total missing values in the whole dataset
total_missing = missing_per_column.sum()
print("\nTotal missing values in dataset:", total_missing)
```

```python
# Q3 (safe version): Replace missing values in Quantity and Price with their mean

quantity_mean = df["Quantity"].mean()
price_mean = df["Price"].mean()

print("Quantity mean:", quantity_mean)
print("Price mean:", price_mean)

# Assign back to the columns explicitly
df["Quantity"] = df["Quantity"].fillna(quantity_mean)
df["Price"] = df["Price"].fillna(price_mean)

print("\nMissing values after filling:")
print(df[["Quantity", "Price"]].isnull().sum())
```

```python
# Q4: Remove rows with missing Product Category or Region

# Before dropping, you can optionally check how many are missing
print("Missing before drop:")
print(df[["Product Category", "Region"]].isnull().sum())

# Drop rows where either column is NaN
df = df.dropna(subset=["Product Category", "Region"])

print("\nMissing after drop:")
print(df[["Product Category", "Region"]].isnull().sum())
```

```python
# Q5: Create Revenue column using NumPy

# df["Revenue"] = np.array(df["Quantity"]) * np.array(df["Price"])
df["Revenue"] = df["Quantity"] * df["Price"]
# Show first few rows to confirm
df.head()
```

```python
# Q6: Total revenue (sum of Revenue column)

total_revenue = df["Revenue"].sum()
print("Total revenue:", round(total_revenue, 2))
```

```python
# Q7: Total revenue by Product Category

revenue_by_category = df.groupby("Product Category")["Revenue"].sum()
revenue_by_category_2d = revenue_by_category.round(2)
print("Revenue by Product Category:")
print(revenue_by_category)
print("Revenue by Product Category with two decimal:")
print(revenue_by_category_2d)
```

```python
# Q8: Top 3 and bottom 3 Product Categories by revenue

# Make sure we use the Series from Q7 (or recompute)
revenue_by_category = df.groupby("Product Category")["Revenue"].sum().round(2)

# Sort descending
revenue_sorted = revenue_by_category.sort_values(ascending=False)

top_3_categories = revenue_sorted.head(3)
bottom_3_categories = revenue_sorted.tail(3)

print("Top 3 categories by revenue:")
print(top_3_categories)

print("\nBottom 3 categories by revenue:")
print(bottom_3_categories)
```

```python
# Q9: Total revenue by Region

revenue_by_region = df.groupby("Region")["Revenue"].sum().round(2)

print("Revenue by Region:")
print(revenue_by_region)

# Region with highest and lowest revenue
highest_region = revenue_by_region.idxmax()
lowest_region = revenue_by_region.idxmin()

print("\nRegion with highest revenue:", highest_region)
print("Region with lowest revenue:", lowest_region)
```

```python
# Q10: Convert Date to datetime with day first (DD-MM-YYYY)
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

# Extract month number (1–12)
df["Month"] = df["Date"].dt.month

# Total monthly revenue
monthly_revenue = df.groupby("Month")["Revenue"].sum().round(2)
print("Monthly total revenue:")
print(monthly_revenue)

# Mean, median, standard deviation of Revenue using NumPy
revenue_values = df["Revenue"].values

revenue_mean = np.mean(revenue_values)
revenue_median = np.median(revenue_values)
revenue_std = np.std(revenue_values)

print("\nRevenue mean:", round(revenue_mean, 2))
print("Revenue median:", round(revenue_median, 2))
print("Revenue standard deviation:", round(revenue_std, 2))
```
